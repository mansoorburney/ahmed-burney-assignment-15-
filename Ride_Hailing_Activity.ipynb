# Ride-Hailing Dataset – Activity Sheet
This notebook answers all five questions using `ride_hailing_dataset.csv`.

**How to use in Google Colab:** Run the cells from top to bottom. When prompted, upload the CSV file provided by your instructor.

```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()  # Upload ride_hailing_dataset.csv
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)

print('Dataset loaded successfully!')
print('Rows and columns:', df.shape)
display(df.head())
```

## Q1. What is the average fare by service type?

```python
average_fare = df.groupby('service_type')['fare'].mean().round(2)
display(average_fare.rename('Average fare').to_frame())
```

## Q2. Which city has the most trips?

```python
trips_by_city = df['city'].value_counts()
display(trips_by_city.rename_axis('City').to_frame('Number of trips'))

if not trips_by_city.empty:
    print('City with the most trips:', trips_by_city.idxmax())
    print('Number of trips:', trips_by_city.max())
```

## Q3. What is the average distance per trip?

```python
average_distance = df['distance_km'].mean()
print('Average distance per trip:', round(average_distance, 2), 'km')
```

## Q4. Compare trip ratings by gender.

```python
average_rating_by_gender = df.groupby('gender')['trip_rating'].mean().round(2)
display(average_rating_by_gender.rename('Average trip rating').to_frame())
```

## Q5. Make a bar chart of service type vs. average fare.

```python
average_fare = df.groupby('service_type')['fare'].mean().sort_values(ascending=False)
display(average_fare.rename('Average fare').to_frame())

average_fare.plot(kind='bar', figsize=(8, 5), rot=0)
plt.title('Average Fare by Service Type')
plt.xlabel('Service type')
plt.ylabel('Average fare')
plt.tight_layout()
plt.show()
```
